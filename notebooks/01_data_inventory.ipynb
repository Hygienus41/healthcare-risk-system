```python
# Import the tools we'll use:

from pathlib import Path
import pandas as pd
import os

DATA_DIR = Path("../data/raw/synthea")

print(f"Data directory: {DATA_DIR.resolve()}")
```

```python
csv_files = sorted(DATA_DIR.glob("*.csv"))

inventory = []

for file in csv_files:
    inventory.append({
        "table": file.stem,
        "file_name": file.name,
        "size_mb": round(file.stat().st_size / (1024 ** 2), 2)
    })

inventory_df = pd.DataFrame(inventory)

inventory_df
```

```python
table_stats = []

for file in csv_files:
    df_sample = pd.read_csv(file, nrows=5)

    with open(file, "rb") as f:
        row_count = sum(1 for _ in f) - 1

    table_stats.append({
        "table": file.stem,
        "rows": row_count,
        "columns": len(df_sample.columns),
        "size_mb": round(file.stat().st_size / (1024 ** 2), 2)
    })

stats_df = pd.DataFrame(table_stats).sort_values("rows", ascending=False)

stats_df
```

```python
# Save the inventory

from pathlib import Path

marts_dir = Path("../data/marts")
marts_dir.mkdir(parents=True, exist_ok=True)

stats_df.to_csv(marts_dir / "table_inventory.csv", index=False)

print("Inventory saved successfully.")
```
